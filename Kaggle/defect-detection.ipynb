```python
import os

dataset_path = '/kaggle/input/datasets/liuxiaolong1/pcb-defect-detection-dataset/PKU-Market-PCB(Data enhanced version)'

data_yaml = f"""
path: {dataset_path}
train: train/images
val: valid/images

names:
  0: Missing_hole
  1: Mouse_bite
  2: Open_circuit
  3: Short
  4: Spur
  5: Spurious_copper
"""

with open('/kaggle/working/data.yaml', 'w') as f:
    f.write(data_yaml)
```

```python
import torch
print(torch.cuda.is_available())
print(torch.cuda.get_device_name(0))
```

```python
from ultralytics import YOLO

model = YOLO('yolov8n.pt')

model.train(
    data='/kaggle/working/data.yaml',
    epochs=30,
    imgsz=640,
    batch=16
)
```

```python
model = YOLO('/kaggle/working/runs/detect/train/weights/best.pt')

model.predict(
    source=dataset_path + '/valid/images',
    save=True
)
```

```python
import os
from ultralytics import YOLO

# Load your trained model
model = YOLO('/kaggle/working/runs/detect/train/weights/best.pt')

# Folder containing images
folder = dataset_path + '/valid/images'   # or any folder

for img in os.listdir(folder):
    path = os.path.join(folder, img)

    # Run prediction
    results = model(path, verbose=False)

    # Extract predictions
    boxes = results[0].boxes

    if boxes is None or len(boxes) == 0:
        status = "OK"
        defect_type = "None"
    else:
        status = "DEFECT"

        # Get class names
        class_ids = boxes.cls.cpu().numpy().astype(int)
        class_names = [model.names[i] for i in class_ids]

        defect_type = ", ".join(set(class_names))  # unique defects

    print(f"{img} → {status} → {defect_type}")
```
